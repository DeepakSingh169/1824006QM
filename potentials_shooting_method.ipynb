# Infinite potential well

```python
def potential(xi):
    return 0

energy = quantum_shooting(potential, 0, 1,30, 50, 0.01,1e-8, 1e-8, 100)

xi_values, psi, psi_normalized = wavefunction(potential, 0, 1, energy, 0.01)

plot_wavefunction(psi, psi_normalized, energy, xi_values, potential_name="infinite_potential_well")
```

# Harmonic Potential

```python

def potential(xi):
    return xi**2

xi_initial = -5
xi_final = 5
h = 0.01

energy_low = 0.5
energy_high = 1.5

energy = quantum_shooting(potential,xi_initial,xi_final,energy_low,energy_high,h,1e-8,1e-8,100)

xi_values, psi, psi_normalized = wavefunction(potential,xi_initial,xi_final,energy,h)

plot_wavefunction(psi,psi_normalized,energy,xi_values,potential_name="Quantum Harmonic Oscillator")
```

# Finite Potential well

```python
def potential(xi):
    if abs(xi) <= 1:
        return -10
    else:
        return 0

xi_initial = -5
xi_final = 5
h = 0.01

energy_low = -9
energy_high = -8

energy = quantum_shooting(potential,xi_initial,xi_final,energy_low,energy_high,h,1e-8,1e-8,100)

xi_values, psi, psi_normalized = wavefunction(potential,xi_initial,xi_final,energy,h)

plot_wavefunction(psi,psi_normalized,energy,xi_values,potential_name="Finite Square Well")
```
