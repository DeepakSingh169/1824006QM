```python
import numpy as np
import matplotlib.pyplot as plt
import math as m
```

```python
a = 1.0
```

```python
def recurrence_relation(n,l):
    j_max = n - l - 1
    c_o   = 1
    coefficients = np.zeros(j_max + 1)
    coefficients[0] = c_o

    for j in range(0,j_max):
        num = 2*(j + l + 1 - n)
        denom = (j + 1)*(j + 2*l + 2)
        coefficients[j+1] = (num*coefficients[j])/denom

    return coefficients
```

```python
def radial_solution(r_min,r_max,num_points,n,l):
    r = np.linspace(max(r_min,1e-12),r_max,num_points)
    rho = r/(a*n)

    coefficients = recurrence_relation(n,l)
    v_rho = np.polyval(coefficients[::-1], rho)

    norm_constant_first = (2/(n*a))**(l + 1.5)
    norm_constant_second = np.sqrt(m.factorial(n - l - 1)/(2*n*m.factorial(n + 1)))
    norm_constant = norm_constant_first*norm_constant_second

    R = norm_constant*(((rho**(l+1))*np.exp(-rho)*v_rho)/r)

    return R,r
```

```python
r_min = 1e-11
r_max = 30
num_points = 1000
```

```python
n_states = [1,2,3,4,5]
l = 0
for states in n_states:
    R,r = radial_solution(r_min,r_max,num_points,states,l)
    plt.plot(r,R,label=f"(n={states},l={l})",linewidth=2)

plt.axhline(0,color="black",linewidth=0.8,linestyle="--")
plt.axvline(0,color="black",linewidth=0.8,linestyle="--")

plt.title("Radial Wave functions of hydrogen atom")
plt.xlabel("Radial Distance r")
plt.ylabel(r"$R_{nl}(r)$")
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```
